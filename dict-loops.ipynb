```python
grades = {'Juan': {'Maths': 2},
          'John': {'History': 1, 'Biology': 3},
          'Gianni': {'Biology': 1}}

grades['Juan']['History'] = 2
grades['Juan']['Biology'] = 1

grades['John']['Maths'] = 1

grades['Gianni']['History'] = 4
grades['Gianni']['Maths'] = 3
print(grades)
```

```python
del grades['Juan']['History']
print(grades["Juan"])
```

```python
# total length of string in list ["red","green","blue"] -> 12
list = ["red","green","blue"]
total = 0
for word in list:
    total = total + len(word)
print(total)
```

```python
# list of word lengths -> [3,5,4]
lengths = []    # empty list to fill in loop
for word in list:
    lengths.append(len(word))    #empty list.append(+length of word
print(lengths)
```

```python
# concatenate all words -> "redgreenblue"
words = ["red","green","blue"]
result = ""
for word in words:
    result = result + word    # like .append for lists
print(result)
```

```python
# cumulative sum -> [1,3,5,10]
cumulative = []    # starting point
data = [1,2,2,5]
total = 0
for number in data:
    total = total + number
    cumulative.append(total)
print(cumulative)
```

```python
# print a if multiple of 3, b if not
message = ""
for number in range(10):
    if(number % 3) == 0:   #% module ie remainer of division
        message = message + "a"
    else:
        message = message + "b"
print(message)
```

```python
import random

def egglabel(mass):
    egglabel = "unlabelled"
    if mass >= 90:
        egglabel = "might be dirty"
    elif mass >= 85:
        egglabel = "jumbo"
    elif mass >= 70:
        egglabel = "large"
    elif mass < 70 and mass >= 55:
        egglabel = "medium"
    elif mass <= 70 and mass >= 50:
        egglabel = "small"
    else:
        egglabel = "too light, might be spoiled"
    return egglabel

for i in range(10):
    mass = 70 + 20.0 * (2.0 * random.random() -1.0)
    print(mass, egglabel(mass))
```

```python
## simulate a dynamical system

#1: defining function
# x = current population at tinme t so x(t)
# r = 1   parameter
def logistic_map(x,r):
    # should return value of state system at time t + 1
    # via f(t+1) = r * f(t) * [1 - f(t)]
    return r * x * (1 -x)
```

```python
# 2: itearte starting from initial population for t_final
# store results in list
# so we have results at t = [0, 1, ...t_final]
initial_pop = 0.5
t_final = 10
r = 1.0
population = [initial_pop]  # not empty, starting at initial

for t in range(t_final):   # 0 to end
    # nope: logistic_map(initial_pop, r)
    population.append(logistic_map(population[t],r))
    # adding list starting at initial
    # run function(x,r)
        # r already defined as 1
        # x = population[t]
    # because we want population in each of the loops

print(population)
```

```python
#3: def functoin(initial_pop, t_final, r)
# returns list of valies at times t 
# run for periods t_final = 100 and 1000

def iterate(initial_pop, t_final, r):
    population = [initial_pop]
    for t in range(t_final):
        population.append(logistic_map(population[t], r))
    return population

for period in (10,100,1000):
    population = iterate(0.5, period, 1)  
        # 0.5 cause inital_pop, period 10/100/100
        # 1 = r
    print(population[-1])
```

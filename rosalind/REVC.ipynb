## GC

#### in class version

```python
# calculate GC content for each sequence
## for loop over all sequences
## make GC calculation a function
def GC(dna):
    G_count = dna.count("G")
    C_count = dna.count("C")
    return (G_count + C_count) / len(dna) * 100
```

```python
# GC("CGCGCGCGGC")
```

```python
# find way to get sample input
sequences= {}

with open("./rosalind_gc.txt", "r") as f:
    for line in f:
        # figure out if header or sequence
        if line[0] == ">":
            lastID = line[1:].strip() # to remove \n at end
            sequences[lastID] = ""  # empty value
        else:
            # sequence line
            sequences[lastID] += line.rstrip()
            

# input file:
## for each line
## each sequence belongs to the last header (>) 
## if > then headerline = sequence ID
### make new dictkey = sequences[this line] = ___ (empty)
## else : sequence = add to dict via sequences[last ID] + 
```

```python
# compare contents and find maximum GC 
## dictionary - ID (key) : content (value)
gc_content = {}
for seq_ID, seq in sequences.items():
    gc_content[seq_ID] = GC(seq)
```

```python
# find corresponding ID for maximum
max_gc = max(gc_content.values())

for seq_ID in gc_content:
    if gc_content[seq_ID] == max_gc:
        print(seq_ID)
        print(round(max_gc, 4))
```

## PROT
turn RNA string into Protein string
given: RNA string s
return: protein string encoded by s

```python
codontable = {}
with open("./codontable.txt", "r") as f:
    for line in f:
      key, value = line.split(" ")
      codontable[key.strip()] = value.strip()
print(codontable)
```

```python
RNA = ""

with open("/Users/sophi/Downloads/rosalind_prot.txt", "r") as input:
    for line in input:
        RNA += line.strip()
# print(RNA)
```

```python
protein_seq = ""   # empty start

for i in range(0, len(RNA), 3):
    """ read RNA in steps of 3
    and add corresponding amino acid into protein_seq """
     # = triplett code
    triplett = RNA[i:i+3]
    # get amino acid = dictionary value
    amino = codontable[triplett]
    if amino == "Stop":
        # ignore
        protein_seq = protein_seq
    else:
        # print amino acid
        protein_seq += amino
        
print(protein_seq)
```

```python
mass = 3.54
if mass > 3.0:
    print("this is too heavy")
if mass <= 4.0:
    print("this fits in the box")
if mass == 3.54:
    print("very specific")
```

```python
masses = [3.54, 5,7,1,0.4,6.68,8,12,5,1,2,0.2,0.3]
total_weight = 0
for m in masses:
    if m < 3.0:
        # put it into trunk
        total_weight = total_weight + m
print("my trunk contains items with total mass of", total_weight)
```

```python
lights = ["green", "red", "yellow", "blue", "green", "green", "yellow"]
for l in lights:
    if l == "green":
        print("keep moving")
    elif l == "yellow":    # elif to not exectue all of the ifs (ie for > conditions)
        print("FASTER")
    elif l == "red":
        print("stop")
    else:
        print("maybe stop")
```

```python
# find largest & smallest in list
values = [-2,1,65,78,-54,-24,100]
smallest, largest = None, None
for v in values:
    if smallest == None and largest == None:
        smallest, largest = v, v
    else:   # keep or new
        smallest = min(smallest, v)
        largest = max(largest, v)
print(smallest, largest)
# with None betters use "is" not "=="
```

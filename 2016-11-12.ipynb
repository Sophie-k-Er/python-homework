#### jigsaw puzzle
rearrange statement so random DNA base and its index is printed

```python
import random # to get module
bases = "ACTTGCTTGAC" # define input
n_bases = len(bases)   # used in randrange so define before that
rand_index = random.randrange(n_bases)   # get random index form len bases
print("random base", bases[rand_index], "base index", rand_index)
    #bases[index] to get base ie position corresponding to random index
```

### dictionaries

```python
d = {"alena":1, "lara":1, "stephanie":1}
print(d.keys())
d.values()
```

```python
d["sophie"] = 1
print(d)
```

```python
student_numbers = {"BioTech" : 16,
                   "CompBio" : 12,
                   "post-gen bio" : 20,
                   "maths" : 0}
print("#number of students", student_numbers)
```

```python
print(student_numbers["CompBio"])
```

```python
student_numbers.update({"maths" : 34})
print(student_numbers["maths"])
```

```python
grades = {'Juan': {'Maths': 2},
          'John': {'History': 1, 'Biology': 3},
          'Gianni': {'Biology': 1}}

grades['Juan']['History'] = 2
grades['Juan']['Biology'] = 1

grades['John']['Maths'] = 1

grades['Gianni']['History'] = 4
grades['Gianni']['Maths'] = 3
print(grades)
```

```python
del grades['Juan']['History']
print(grades["Juan"])
```

```python
a = {"name":"John", "age":20}
b = {"name":"May", "age":23}
customers = {"c1":a, "c2" : b} # c1 John c2 may
# for printing name may
print(customers["c2"]["name"])
# dictname[key][key nested]
```

### for loops

```python
for number in [2,3,4,5]:
	print("current number is", str(number))
```

```python
for number in [2,3,4,5,6,7,8,9,10]:
	print("The current number is", number)
```

```python
primes = [2,3,5,7]
for number in primes:   # can name it whatever i want
    square = number**2
    cube = number**3
    print(number, square, cube)
```

## GC
GC content is percentage of GC in string
format: >Rosalind_XXXX \n string
given: DNA strings in FASTA format
return: ID of string having highest GC conent + GC content %

```python
# "C:\Users\sophi\Downloads\rosalind_gc.txt"
samples = {}

with open("./rosalind_gc.txt", "r") as f:
    for line in f:
        # smth f.readline ??
      key, value = line.split("\n")
      samples[key.strip()] = value.strip()
    # split at first " " into key value # for all lines
    # creates dict samples
print(samples.values())
print(samples.keys())
```

```python
# function over samples
# count length, count CG -> calulate percentage
# count which string higest CG & print
# dict?

# sample = "CCACCCTCGTGGTATGGCTAGGCATTCAGGAACCGGAGAACGCTTCAGACCAGCCCGGACTGGGAACCTGCGGGCAGTAGGTGGAAT"

def GC_content(DNA):
    G_count = DNA.count("G")
    C_count = DNA.count("C")
    GC_count = G_count + C_count
    GC_content = (GC_count/ len(DNA)) * 100
    # print(GC_content)

# run GC_content function on values          
for DNA in samples.keys():
    GC_content(DNA)
    # print(GC_content)
```

```python
# get maximum GC_content
maxkey = None
maxvalue = 0

for k,v in samples.items():
    if v > maxvalue:
        maxvalue = v
        maxkey = k
        
print(maxkey, "\n", maxvalue)  
# print(ID, CG_content)
```

## FIB

recurrence relation
- defines a sequence where each term is given as a function of one or more of its preceding terms
- can be solved via
  - substitution method
  - recurrence trees
  - master theorem
general form: T(n) = a * T(n/b) + f(n)
- T(n) is function were defining
- a is number of subproblems in recurrence
- n/b is size of each subproblem
- f(n) is cost outsie recursive calls

rules:
- there is 1 initial pair of baby rabbits
- month 1 and 2 both have 1 pair of rabbits (baby, then grown)
- every month add k pairs of baby rabbits for each pair of grown rabbits
- baby rabbits need one month to be grown
    - meaning: first baby rabbits in month 3
- we keep going of n months

plan:
- for each month in n:
  - aus adult wird adult + k * baby: 
      - for pair in list grown add k pairs in list baby
    - for each pair in list baby add one pair in list grown
  - then remove pairs from list baby
- total pairs = grown + baby

- make function with starting point input
  - that function moves whats written above
  - as longs as current month != last month keep doing that
    - meaning: call that function within function
